<h1 align="center">McCulloch-Pitts: The First Computational Neuron</h1>

The **McCulloch–Pitts neuron** is one of the earliest mathematical models designed to simulate the behavior of biological neurons in the brain.  
It was introduced in **1943** by *Warren McCulloch* and *Walter Pitts*.  
Their goal was to propose a simple computational model that could describe how biological neurons process information and make decisions based on input signals.

- The model receives multiple inputs, each associated with a weight.  
- The weighted sum of these inputs is compared against a threshold.  
- If the sum exceeds the threshold, the neuron outputs **1**; otherwise, it outputs **0**.  
- This model laid the groundwork for artificial neural networks and demonstrated that simple neurons can implement logical functions such as **AND** and **OR**.

<h2 align="left">AND Gate</h2>

In the McCulloch–Pitts model, the **AND gate** can be implemented by assigning the weights  
- **w₁ = w₂ = 1** and a threshold **θ = 2**.  
- The neuron fires (**output = 1**) only when both inputs are active.

<div align="center">
    <img src="andgat.jpg" width="800">
</div>


```python
# AND Gate using McCulloch-Pitts neuron
def AND(x1, x2):
    
    theta = 2
    s = x1 + x2
    
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {AND(x1, x2)}")
```

<h2 align="left">OR Gate</h2>

In the McCulloch–Pitts model, the **OR gate** can be implemented by assigning the weights  
- **w₁ = w₂ = 1** and a threshold **θ = 1**.  
- The neuron fires (**output = 1**) when at least one of the inputs is active.

<div align="center">
    <img src="orgate.jpg" width="800">
</div>


```python
# OR Gate using McCulloch-Pitts neuron
def OR(x1, x2):
    theta = 1
    s = x1 + x2
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {OR(x1, x2)}")
```

<h2 align="left">AND-NOT Gate</h2>

In the **McCulloch–Pitts** model, this neuron implements the **AND‑NOT** function by assigning the weights  
- **w₁ = 2** and **w₂ = −1**, with a threshold **θ = 2**.  
- The neuron produces an output of **1** only when **x₁ = 1** and **x₂ = 0**.

<div align="center">
    <img src="andnot.jpg" width="800">
</div>


```python
# AND Not Gate using McCulloch-Pitts neuron
def AND_Not(x1, x2):
    w1, w2 = 2, -1
    theta = 2
    s = w1*x1 + w2*x2
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {AND_Not(x1, x2)}")
```

<h2 align="left">XOR Gate</h2>

In the **McCulloch–Pitts** model, the XOR gate cannot be implemented using a single neuron.  
Instead, it requires a **multi‑layer network** with two hidden units:

1. **Hidden Neuron 1:** computes **x₁ AND NOT x₂**  
2. **Hidden Neuron 2:** computes **x₂ AND NOT x₁**  
3. **Output Neuron:** applies an **OR** operation to the outputs of the two hidden neurons

This structure correctly produces a XOR behavior:  
the output is **1** only when exactly one of the inputs is active.

<div align="center">
    <img src="xor.jpg" width="800">
</div>


```python
# Activation function (McCulloch-Pitts)
def step(y_in, theta=2):
    return 1 if y_in >= theta else 0


# Hidden neuron Z1 = x1 AND NOT x2
def z1(x1, x2):
    y_in = 2*x1 + (-1)*x2
    return step(y_in)


# Hidden neuron Z2 = x2 AND NOT x1
def z2(x1, x2):
    y_in = 2*x2 + (-1)*x1
    return step(y_in)


# Output neuron Y = Z1 OR Z2
def XOR(x1, x2):
    Z1 = z1(x1, x2)
    Z2 = z2(x1, x2)

    y_in = 2*Z1 + 2*Z2
    return step(y_in)


# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {XOR(x1, x2)}")
```

<h2 align="left">Half Adder</h2>

A **Half Adder** is a basic **combinational logic circuit** that performs the addition of two single‑bit binary numbers.  
It produces two outputs:

- **Sum (S):** Computed using an **XOR gate**  
- **Carry (C):** Computed using an **AND gate**

<div align="center">
    <img src="HA.jpg" width="800">
</div>


```python
# Half Adder using McCulloch-Pitts neuron

def Half_Adder(A, B):
    s = XOR(A, B)
    c = AND(A, B)
    return s, c

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("A  B  |  S  C ")
    print("-------------")
    
    for A in [0,1]:
        for B in [0,1]:
            s,c= Half_Adder(A, B)
            
            print(f" {A}  {B}  | {s}  {c}")

```

<div style="border:1px solid #d0d7de; padding:16px; border-radius:8px;">
<h2 align="left">Full Adder</h2>

A **Full Adder** is a combinational logic circuit that adds three input bits:  
two significant bits (**A** and **B**) and a **carry‑in (Cin)** from the previous stage.

The circuit produces two outputs:

- **Sum (S):** computed using XOR operations  
- **Carry Out (Cout):** generated using a combination of AND and OR gates

A multi‑bit adder can be constructed by **connecting several 1‑bit Full Adders in series**, allowing the carry signal to propagate from one stage to the next.  
Each logic gate used in the circuit (**AND, OR, XOR**) can be modeled using **McCulloch–Pitts neurons**, illustrating how simple neuron models can be combined to perform more complex computational tasks such as multi‑bit binary addition.

<div align="center">
    <img src="fa.jpg" width="800">
</div>


```python
# Full Adder using McCulloch-Pitts neuron
 
def Full_Adder(A, B, c_in):
    s, c = Half_Adder(A, B)
    S, C = Half_Adder(s, c_in)
    C_out = OR(c, C)
    return S, C_out

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("c_in  A  B | Sum  C_out ")
    print("----------------------")
    
    for c_in in [0,1]:
        for A in [0,1]:
            for B in [0,1]:
                S, C_out = Full_Adder(A, B, c_in)
                print(f"{c_in}   {A}   {B}  |   {S}   {C_out}")
```

<h2 align="left">Multiplier</h2>

The **binary multiplication** of two 2‑bit operands produces results in the range **0 to 9 (decimal)**.  
The maximum product, **9**, corresponds to the binary representation **1001**.

A 2‑bit binary multiplier can be implemented using:
- **AND gates** to generate partial products  
- **Half Adders** and **Full Adders** to sum these partial products  
- **Carry propagation** between stages, resulting in a 4‑bit output (C₃C₂C₁C₀)

This demonstrates how basic logic primitives—when combined—can perform complex arithmetic operations, each step representable by simple **McCulloch–Pitts neurons**.

<div align="center">
    <img src="multiplier.jpg" width="800">
</div>


```python
# --- MULTIPLIER 2-bit ---
def binary_multiplier(A1, A0, B1, B0):
    # Partial products
    P0 = AND(A0, B0)
    P1 = AND(A0, B1)
    P2 = AND(A1, B0)
    P3 = AND(A1, B1)

    # First Half Adder: P1 + P2
    S1, carry1 = Half_Adder(P1, P2)

    # Second Half Adder: P3 + carry1
    S2, carry2 = Half_Adder(P3, carry1)

    # Output bits (standard binary order)
    C0 = P0
    C1 = S1
    C2 = S2
    C3 = carry2

    return C3, C2, C1, C0


# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("A1 A0 | B1 B0 | C3 C2 C1 C0 ")
    print("-----------------------------")

    for A1 in [0, 1]:
        for A0 in [0, 1]:
            for B1 in [0, 1]:
                for B0 in [0, 1]:
                    C3, C2, C1, C0 = binary_multiplier(A1, A0, B1, B0)
                    print(f"{A1}  {A0} | {B1}  {B0} |  {C3}  {C2}  {C1}  {C0}")
```

<h2 align="left">Modeling the Perception of Hot and Cold</h2>

This project models basic **heat** and **cold** perception using a simple **McCulloch–Pitts neural network**.  
The system receives two binary inputs:

- **X₁:** heat stimulus  
- **X₂:** cold stimulus  

It produces two binary outputs:

- **Y₁:** perceived heat  
- **Y₂:** perceived cold  

To keep the model biologically plausible and logically consistent, it assumes that **heat and cold do not occur simultaneously**, which limits the valid input states to:

- (0, 0): no stimulus  
- (1, 0): heat  
- (0, 1): cold  

The network uses positive and inhibitory connections to simulate how short or long exposure to a stimulus may shift the perceived response over time.

<div align="center">
    <img src="hotandcold.jpg" width="450">
</div>


```python
# McCulloch–Pitts Neuron
def McCullochPitts_neuron(inputs, weights):
    threshold=1
    #for example inputs=[i1,12] & weights=[w1,w2] -> total = i1*w1 + i2*w2 & 
    #list(zip(inputs, weights))=[(i1, w1), (i2, w2)]
    
    total = sum(i * w for i, w in zip(inputs, weights))
    return 1 if total >= threshold else 0


# McCulloch–Pitts Network (Based on Image)

def McCullochPitts_network(X1, X2):
    
    # ---- Hidden Layer ----
    # Z2 = step( 2*X2 )
    Z2 = McCullochPitts_neuron([X2], [2])

    # Z1 = step( (-1)*X2 + 2*Z2 )
    Z1 = McCullochPitts_neuron([X2, Z2], [-1, 2])

    # ---- Outputs ----
    # Y1 = step( 2*X1 + 2*Z1 )
    Y1 = McCullochPitts_neuron([X1, Z1], [2, 2])

    # Y2 = step( 1*Z2 + 1*X2 )
    Y2 = McCullochPitts_neuron([Z2, X2], [1, 1])

    return Z1, Z2, Y1, Y2



# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1 X2 | Z1 Z2 | Y1 Y2")
    print("-----------------------")
    
    for X1 in [0, 1]:
        for X2 in [0, 1]:
            Z1, Z2, Y1, Y2 = McCullochPitts_network(X1, X2)
            print(f"{X1}  {X2}  |  {Z1}  {Z2}  |  {Y1}  {Y2}")
```
