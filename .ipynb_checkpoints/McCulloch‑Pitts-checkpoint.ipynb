<h2 style="text-align">AND gate :</h2>
<b>
In the McCulloch-Pitts model, the AND gate can be implemented by assigning weights 𝑤1=𝑤2=1 and threshold θ=2. The neuron fires only when both inputs are active.
</b>
<div style="text-align:center;">
    <img src="andgat.jpg" width="1000">
</div>


```python
# AND Gate using McCulloch-Pitts neuron
def AND(x1, x2):
    
    theta = 2
    s = x1 + x2
    
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {AND(x1, x2)}")
```

<h2 style="text-align">OR gate :</h2>
<b>In the McCulloch-Pitts model, the OR gate can be implemented by assigning weights 𝑤1=𝑤2=2 and threshold θ=2. The neuron fires when at least one of the inputs is active.</b>
<div style="text-align:center;">
    <img src="orgate.jpg" width="1000">
</div>

```python
# OR Gate using McCulloch-Pitts neuron
def OR(x1, x2):
    theta = 1
    s = x1 + x2
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {OR(x1, x2)}")
```

<h2 style="text-align">AND Not gate :</h2>
<b>In the McCulloch‑Pitts model, this neuron implements the AND‑NOT function by assigning weights 𝑤1=2 and 𝑤2=−1 with threshold θ=2, producing an output of 1 only when 𝑥1=1 and 𝑥2=0.</b>
<div style="text-align:center;">
    <img src="andnot.jpg" width="1000">
</div>

```python
# AND Not Gate using McCulloch-Pitts neuron
def AND_Not(x1, x2):
    w1, w2 = 2, -1
    theta = 2
    s = w1*x1 + w2*x2
    return 1 if s >= theta else 0

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {AND_Not(x1, x2)}")
```

<h2 style="text-align">XOR gate :</h2>
<b>In the McCulloch‑Pitts model, the XOR gate is implemented using two hidden neurons that compute 𝑥1 AND NOT 𝑥2 and 𝑥2 AND NOT 𝑥1, whose outputs are then combined with an OR neuron to produce the final output.</b>
<div style="text-align:center;">
    <img src="xor.jpg" width="1000">
</div>

```python
# Activation function (McCulloch-Pitts)
def step(y_in, theta=2):
    return 1 if y_in >= theta else 0


# Hidden neuron Z1 = x1 AND NOT x2
def z1(x1, x2):
    y_in = 2*x1 + (-1)*x2
    return step(y_in)


# Hidden neuron Z2 = x2 AND NOT x1
def z2(x1, x2):
    y_in = 2*x2 + (-1)*x1
    return step(y_in)


# Output neuron Y = Z1 OR Z2
def XOR(x1, x2):
    Z1 = z1(x1, x2)
    Z2 = z2(x1, x2)

    y_in = 2*Z1 + 2*Z2
    return step(y_in)


# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1  X2 |  Y ")
    print("-------------")
    
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(f"{x1}   {x2}  |  {XOR(x1, x2)}")
```

<h2 style="text-align">Half Adder :</h2>
<b>
 A Half Adder is a basic combinational circuit that adds two binary bits and produces two outputs: Sum (computed using an XOR gate) and Carry (computed using an AND gate).
</b>
<div style="text-align:center;">
    <img src="HA.jpg" width="1000">
</div>

```python
# Half Adder using McCulloch-Pitts neuron

def Half_Adder(A, B):
    s = XOR(A, B)
    c = AND(A, B)
    return s, c

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("A  B  |  S  C ")
    print("-------------")
    
    for A in [0,1]:
        for B in [0,1]:
            s,c= Half_Adder(A, B)
            
            print(f" {A}  {B}  | {s}  {c}")

```

<h2 style="text-align">Full Adder :</h2>
<b>
A multi-bit full adder circuit can be constructed by connecting multiple basic full adders in series. Each 1-bit full adder is built using interconnected logic gates (AND, OR, XOR), which themselves can be modeled with simple McCulloch–Pitts neurons. This approach demonstrates how simple neurons can be combined to perform complex multi-bit binary addition with carry propagation.
</b>
<div style="text-align:center;">
    <img src="fa.jpg" width="1000">
</div>

```python
# Full Adder using McCulloch-Pitts neuron
 
def Full_Adder(A, B, c_in):
    s, c = Half_Adder(A, B)
    S, C = Half_Adder(s, c_in)
    C_out = OR(c, C)
    return S, C_out

# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("c_in  A  B | Sum  C_out ")
    print("----------------------")
    
    for c_in in [0,1]:
        for A in [0,1]:
            for B in [0,1]:
                S, C_out = Full_Adder(A, B, c_in)
                print(f"{c_in}   {A}   {B}  |   {S}   {C_out}")
```

<h2 style="text-align">Multiplier :</h2>
<b>
The binary multiplication of two 2‑bit operands produces results in the range 0 to 9 (decimal), where the maximum product, 9, corresponds to the binary representation 1001.
</b>
<div style="text-align:center;">
    <img src="multiplier.jpg" width="1000">
</div>

```python
# --- MULTIPLIER 2-bit ---
def binary_multiplier(A1, A0, B1, B0):
    # Partial products
    P0 = AND(A0, B0)
    P1 = AND(A0, B1)
    P2 = AND(A1, B0)
    P3 = AND(A1, B1)

    # First Half Adder: P1 + P2
    S1, carry1 = Half_Adder(P1, P2)

    # Second Half Adder: P3 + carry1
    S2, carry2 = Half_Adder(P3, carry1)

    # Output bits (standard binary order)
    C0 = P0
    C1 = S1
    C2 = S2
    C3 = carry2

    return C3, C2, C1, C0


# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("A1 A0 | B1 B0 | C3 C2 C1 C0 ")
    print("-----------------------------")

    for A1 in [0, 1]:
        for A0 in [0, 1]:
            for B1 in [0, 1]:
                for B0 in [0, 1]:
                    C3, C2, C1, C0 = binary_multiplier(A1, A0, B1, B0)
                    print(f"{A1}  {A0} | {B1}  {B0} |  {C3}  {C2}  {C1}  {C0}")
```

<h2 style="text-align">Modeling the Perception of Hot and Cold :</h2>
<b>
This project models basic heat and cold perception using a McCulloch-Pitts neural network. The system takes two binary inputs, 𝑋1 (heat) and 𝑋2(cold), and produces two outputs, 𝑌1 and 𝑌2, representing the perception of heat and cold respectively. The model assumes heat and cold stimuli are not simultaneous, limiting valid input states to (0,0), (1,0), and (0,1).
</b>
<div style="text-align:center;">
    <img src="hotandcold.jpg" width="500">
</div>

```python
# McCulloch–Pitts Neuron
def McCullochPitts_neuron(inputs, weights):
    threshold=1
    #for example inputs=[i1,12] & weights=[w1,w2] -> total = i1*w1 + i2*w2 & 
    #list(zip(inputs, weights))=[(i1, w1), (i2, w2)]
    
    total = sum(i * w for i, w in zip(inputs, weights))
    return 1 if total >= threshold else 0


# McCulloch–Pitts Network (Based on Image)

def McCullochPitts_network(X1, X2):
    
    # ---- Hidden Layer ----
    # Z2 = step( 2*X2 )
    Z2 = McCullochPitts_neuron([X2], [2])

    # Z1 = step( (-1)*X2 + 2*Z2 )
    Z1 = McCullochPitts_neuron([X2, Z2], [-1, 2])

    # ---- Outputs ----
    # Y1 = step( 2*X1 + 2*Z1 )
    Y1 = McCullochPitts_neuron([X1, Z1], [2, 2])

    # Y2 = step( 1*Z2 + 1*X2 )
    Y2 = McCullochPitts_neuron([Z2, X2], [1, 1])

    return Z1, Z2, Y1, Y2



# Test All Input Combinations
if __name__ == "__main__":
    print("Truth Table:\n")
    print("X1 X2 | Z1 Z2 | Y1 Y2")
    print("-----------------------")
    
    for X1 in [0, 1]:
        for X2 in [0, 1]:
            Z1, Z2, Y1, Y2 = McCullochPitts_network(X1, X2)
            print(f"{X1}  {X2}  |  {Z1}  {Z2}  |  {Y1}  {Y2}")
```

## Conclusion
<b>
In this notebook, we explored the McCulloch–Pitts neuron — the earliest mathematical model of an artificial neuron and one of the foundational concepts behind modern artificial intelligence.
We demonstrated how a single threshold neuron can simulate fundamental logical operations such as AND and OR, and how combining simple neurons enables the construction of more complex computational systems like a Full Adder. These experiments illustrate how binary logic and weighted summation form the backbone of neural computation.
However, we also observed a critical limitation: a single-layer McCulloch–Pitts neuron cannot solve non-linearly separable problems such as XOR. This limitation historically motivated the development of multi-layer neural networks, ultimately leading to modern deep learning architectures.
Understanding these roots provides valuable insight into how today’s large-scale neural systems evolved from simple threshold logic models.
